```python
from gosdt import GOSDTClassifier, ThresholdGuessBinarizer
from gosdt._tree import Leaf
```

```python
thresholder = ThresholdGuessBinarizer()
X_bin = thresholder.fit_transform(X, y)

gosdt = GOSDTClassifier(regularization=0.03, depth_budget=5, balance=True)
```

```python
cross_val_score(gosdt, X_bin, y, cv=3, scoring='roc_auc').mean()
```

```python
gosdt.fit(X_bin, y)
```

```python
def gosdt_to_dict(node):
    '''
    Convert a GOSDT tree to a dictionary representation.
    This dictionary representation allows constructing
    tree classifier wrappers from the 
    tree_classifier_wrapper.py file.
    '''
    dict_tree = {}
    ####################
    #features needed for dnf_trees: 
    dict_tree['reference'] = 'true' #queries assumed to always correspond to whether a feature is True; 
                                    # since we're expecting to be dealing with binarized data always with
                                    # thresholds of 0, where left/true corresponds to > 0, this should always hold
    dict_tree['relation'] = '=='
    ####################
    if isinstance(node, Leaf):
        dict_tree['prediction'] = node.prediction
        return dict_tree
    else: 
        dict_tree['feature'] = node.feature
        dict_tree['true'] = gosdt_to_dict(node.left_child)
        dict_tree['false'] = gosdt_to_dict(node.right_child)
        return dict_tree

def print_tree(node, colnames, labels_dict, depth=0, feature_dict=None):
    """
    Recursively prints the tree's information as a string.
    
    Args:
        node (dict): Current node in the tree.
        depth (int): Depth of the current node in the tree (used for indentation).
        colnames (list): List of column names.
        labels_dict (dict): Dictionary mapping the labels in the tree to human-readable labels.
        depth (int): Depth of the current node in the tree (used for indentation).
    
    Returns:
        str: A string representation of the tree.
    """
    if feature_dict is None:
        feature_map = lambda x: x 
    else:
        feature_map = lambda x: feature_dict[x]

    indent = "  " * depth  # Indentation based on depth
    result = ""

    # Check if the current node is a leaf
    if "feature" in node:
        # Internal node
        feature = feature_map(colnames[node["feature"]])
        # Recur on children if they exist
        if "false" in node:
            if "<=" in feature:
                result += f"{indent}{feature.replace("<=", ">")}\n"
            else:
                result += f"{indent}{feature} is False\n"
            result += print_tree(node["false"], colnames, labels_dict, depth + 2, feature_dict)
        if "true" in node:
            if "<=" in feature:
                result += f"{indent}{feature}\n"
            else:
                result += f"{indent}{feature} is True \n"
            result += print_tree(node["true"], colnames, labels_dict, depth + 2, feature_dict)
    else:
        # Leaf node
        result += f"{indent}Leaf: predict {node['prediction']}\n"

    return result

gosdt_dict = gosdt_to_dict(gosdt.trees_[0].tree)
```

```python
colnames = gosdt.feature_names_
labels_dict = {0: 'Survived', 1: 'Died'}
```

```python
thresholded_names = thresholder.get_feature_names_out()
feature_dict = {f'x{i}': name for i, name in enumerate(thresholded_names)}
```

```python
print(print_tree(gosdt_dict, colnames, labels_dict, feature_dict=feature_dict))
```
